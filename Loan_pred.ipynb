```python
import pandas as pd
```

```python
data = pd.read_csv('loan_approval_dataset.csv')
```

```python
data.columns = data.columns.str.strip()
```

```python
data['Assets'] = data.residential_assets_value + data.commercial_assets_value + data.luxury_assets_value + data.bank_asset_value
```

```python
data.drop(columns = ['residential_assets_value', 'commercial_assets_value', 'luxury_assets_value', 'bank_asset_value'], inplace = True)
```

```python
def clean_data(st):
    st = st.strip()
    return st
```

```python
data['education'] = data['education'].map({'Graduate': 1, 'Not Graduate': 0}).fillna(0).astype(int)
```

```python
data.drop(columns = ['loan_id'], inplace = True)
```

```python
data.self_employed = data.self_employed.map({'Yes': 1, 'No': 0}).fillna(0).astype(int)
```

```python
data.loan_status.unique()
```

```python
data.loan_status = data.loan_status.apply(clean_data)
```

```python
data.loan_status = data.loan_status.map({'Approved': 1, 'Rejected': 0}).fillna(0).astype(int)
```

```python
from sklearn.model_selection import train_test_split 
```

```python
input_data = data.drop(columns = ['loan_status'])
output_data = data['loan_status']
```

```python
input_data
```

```python
output_data
```

```python
x_train, x_test, y_train, y_test = train_test_split(input_data, output_data, test_size = 0.2)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape,
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
```

```python
x_test_scaled = scaler.fit_transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

model = LogisticRegression()

```python

model = LogisticRegression()
```

```python
model.fit(x_train_scaled, y_train)
```

```python
model.score(x_test_scaled, y_test)
```
